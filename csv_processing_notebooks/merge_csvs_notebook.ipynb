```python
import pandas as pd

def merge_csvs(file_paths, output_csv=None, fill_value=pd.NA, infer_dtypes=True):
    """
    Merge multiple CSVs into a single DataFrame while:
      - Keeping columns in the order they are first encountered across files.
      - Appending any new columns (from later files) after the existing columns.
    """
    columns_seen = []
    frames = []

    for p in file_paths:
        df = pd.read_csv(p, dtype=None if infer_dtypes else str)
        for c in df.columns:
            if c not in columns_seen:
                columns_seen.append(c)
        frames.append(df)

    reindexed = [f.reindex(columns=columns_seen) for f in frames]
    merged = pd.concat(reindexed, ignore_index=True, sort=False)
    if fill_value is not None:
        merged = merged.fillna(fill_value)
    if output_csv:
        merged.to_csv(output_csv, index=False)
    return merged
```

```python
# Example usage:
sample_paths = ['merged_csv_batch_1.csv', 'merged_csv_batch_2.csv', 'merged_csv_batch_3.csv', 'merged_csv_batch_4.csv', 'merged_csv_batch_5.csv', 'merged_csv_batch_6.csv', 'merged_csv_batch_7.csv']
merged = merge_csvs(sample_paths, output_csv="merged_ig_leads.csv", fill_value="")
print("Merged saved to merged.csv — columns (in first-seen order):", merged.columns.tolist())
merged.head(10)
```
