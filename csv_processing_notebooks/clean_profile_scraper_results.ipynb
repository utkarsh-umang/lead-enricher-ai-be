```python
import pandas as pd

# Path to your CSV
csv_path = "newest.csv"

# Columns you want to KEEP
allowed_cols = [
    "inputUrl", "ownerFullName", "commentsCount", "externalUrl", 
    "followersCount", "followsCount", "fullName",
    "latestPosts/0/likesCount", "latestPosts/1/likesCount",
    "latestPosts/2/likesCount", "latestPosts/3/likesCount",
    "latestPosts/4/likesCount",
    "likesCount", "timestamp", "type", "url", "username"
]

# Load CSV
df = pd.read_csv(csv_path)

# Drop columns NOT in allowed list
df_clean = df[allowed_cols]  # automatically keeps only allowed columns

# Save cleaned CSV
df_clean.to_csv("cleaned_output_3.csv", index=False)

df_clean.head()
```
