```python
import csv

# ---------- CONFIG ----------
HEADER1 = [
    "Business", "Email", "Street", "City", "State", "Postcode", "Country",
    "Phone", "Website", "Youtube", "Contact First Name", "Contact Last Name",
    "Contact Role", "Contact Email"
]

HEADER2 = [
    "Name", "Email", "Phone", "Position", "Company", "Address", "Url"
]

HEADER1_KEYS = {"Business", "Street", "Postcode"}
HEADER2_KEYS = {"Name", "Position", "Url"}

INPUT_FILE = "/Users/utkarshumang/Downloads/merge_csv_batch_8.csv"
OUTPUT1 = "output_1.csv"
OUTPUT2 = "output_2.csv"
# ----------------------------


def normalize(row):
    return [col.replace("ï»¿", "").strip() for col in row]


def is_header(row, header_keys):
    return len(header_keys.intersection(set(row))) >= 2


def split_tables():
    table1_rows = []
    table2_rows = []

    current_table = None

    with open(INPUT_FILE, "r", encoding="utf-8") as f:
        reader = csv.reader(f)
        for row in reader:
            row = normalize(row)

            if not row or all(col == "" for col in row):
                continue

            # ---- Header detection (anywhere) ----
            if is_header(row, HEADER1_KEYS):
                current_table = 1
                continue

            if is_header(row, HEADER2_KEYS):
                current_table = 2
                continue

            # ---- Write data rows only ----
            if current_table == 1:
                padded = row[:len(HEADER1)] + [""] * (len(HEADER1) - len(row))
                table1_rows.append(padded)

            elif current_table == 2:
                padded = row[:len(HEADER2)] + [""] * (len(HEADER2) - len(row))
                table2_rows.append(padded)

    # ---- Write CSVs ----
    with open(OUTPUT1, "w", newline="", encoding="utf-8") as out1:
        writer = csv.writer(out1)
        writer.writerow(HEADER1)
        writer.writerows(table1_rows)

    with open(OUTPUT2, "w", newline="", encoding="utf-8") as out2:
        writer = csv.writer(out2)
        writer.writerow(HEADER2)
        writer.writerows(table2_rows)

    print("✔ Split complete")
    print(f"→ {OUTPUT1}")
    print(f"→ {OUTPUT2}")


if __name__ == "__main__":
    split_tables()
```
