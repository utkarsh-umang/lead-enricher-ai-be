```python
def authenticate_google_sheets():
    """Authenticate and return Google Sheets service"""
    try:
        creds = service_account.Credentials.from_service_account_file(
            CREDENTIALS_FILE, 
            scopes=['https://www.googleapis.com/auth/spreadsheets']
        )
        service = build('sheets', 'v4', credentials=creds)
        return service
    except Exception as e:
        print(f"Error authenticating Google Sheets: {e}")
        return None

def initialize_openai():
    """Initialize OpenAI client"""
    try:
        client = OpenAI(api_key=OPENAI_API_KEY)
        return client
    except Exception as e:
        print(f"Error initializing OpenAI: {e}")
        return None

def count_words(text):
    """Count words in a text string"""
    if not text or text.strip() == "":
        return 0
    return len(text.strip().split())

def collect_datapoints(row):
    """Collect all datapoints from a row with their rankings"""
    datapoints = []
    
    for category, col_index in category_indices.items():
        if category == "SHOP":  # Exclude SHOP from processing
            continue
            
        if col_index < len(row):
            content = str(row[col_index]).strip()
            word_count = count_words(content)
            
            # Only include datapoints with more than 10 words
            if word_count > 10:
                datapoints.append({
                    'category': category,
                    'content': content,
                    'rank': category_ranking[category],
                    'word_count': word_count
                })
    
    # Sort by rank (ascending order - lower rank = higher priority)
    datapoints.sort(key=lambda x: x['rank'])
    return datapoints

def assign_datapoints_to_emails(datapoints):
    """Assign datapoints to emails based on ranking with reuse logic for subsequences"""
    email_assignments = {
        'Email 1': None,
        'Subsequence 1': None,
        'Subsequence 2': None,
        'Subsequence 3': None,
        'Subsequence 4': None
    }
    email_order = ['Email 1', 'Subsequence 1', 'Subsequence 2', 'Subsequence 3', 'Subsequence 4']
    
    # If no datapoints available, return empty assignments
    if not datapoints:
        return email_assignments
    
    # Step 1: Assign unique datapoints first (as many as available)
    for i, email_type in enumerate(email_order):
        if i < len(datapoints):
            email_assignments[email_type] = datapoints[i]
    
    # Step 2: For subsequences 2, 3, 4 - if they don't have a datapoint, reuse first one
    subsequences_to_check = ['Subsequence 2', 'Subsequence 3', 'Subsequence 4']
    for email_type in subsequences_to_check:
        if email_assignments[email_type] is None:
            # Check if any datapoint is available to reuse
            if datapoints:
                # Reuse the first available datapoint (highest ranked)
                email_assignments[email_type] = datapoints[0]
    
    return email_assignments

def get_datapoint_for_email_type(datapoints, email_type):
    """Get the appropriate datapoint for a specific email type"""
    # Use the same logic as assign_datapoints_to_emails but for single email
    email_datapoint_mapping = {
        'Email 1': 0,        # Best datapoint
        'Subsequence 1': 1,  # Second best
        'Subsequence 2': 2,  # Third best
        'Subsequence 3': 3,  # Fourth best
        'Subsequence 4': 4   # Fifth best
    }
    
    datapoint_index = email_datapoint_mapping.get(email_type)
    if datapoint_index is not None and datapoint_index < len(datapoints):
        return datapoints[datapoint_index]
    elif datapoints:  # If no specific datapoint, reuse first one
        return datapoints[0]
    return None

def generate_email_with_datapoint(email_type, category, datapoint_content, first_name, company_name, client):
    """Generate email using OpenAI for emails with datapoints"""
    try:
        # Handle different template structures
        if email_type in ["Email 1", "Subsequence 1"]:
            template = prompt_templates[email_type][category]
        else:
            # For Subsequence 2, 3, 4 - they have simple string templates
            template = prompt_templates[email_type]
        
        prompt = f"""
        {BASE_PROMPT}
        
        Template to use:
        {template}
        
        Person's first name: {first_name}
        Company name: {company_name}
        Datapoint content to personalize with: {datapoint_content}
        Category: {category}
        
        Generate the email following the template exactly, only replacing the bracketed placeholders.
        """
        
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", "content": prompt}],
            max_tokens=500,
            temperature=0.7
        )
        
        return response.choices[0].message.content.strip()
    
    except Exception as e:
        print(f"Error generating email with datapoint: {e}")
        return f"Error generating {email_type}"

def generate_neutral_email(email_type, first_name, company_name, website="", industry="", client=None):
    """Generate neutral email for emails without datapoints"""
    try:
        if email_type in ["Email 1", "Subsequence 1"]:
            template = prompt_templates[email_type]["NEUTRAL"]
            
            if client:
                prompt = f"""
                {BASE_PROMPT}
                
                Template to use:
                {template}
                
                Person's first name: {first_name}
                Company name: {company_name}
                
                Generate the email following the template exactly, only replacing the bracketed placeholders.
                """
                
                response = client.chat.completions.create(
                    model="gpt-4o-mini",
                    messages=[{"role": "user", "content": prompt}],
                    max_tokens=500,
                    temperature=0.7
                )
                
                return response.choices[0].message.content.strip()
            else:
                # Simple replacement for neutral templates
                email = template.replace("[FIRST NAME]", first_name)
                email = email.replace("[COMPANY NAME]", company_name)
                return email.strip()
        
        elif email_type in ["Subsequence 2", "Subsequence 3", "Subsequence 4"]:
            # No neutral templates provided for these subsequences
            return "NO template provided"
        
        return f"Neutral template for {email_type}"
    
    except Exception as e:
        print(f"Error generating neutral email: {e}")
        return f"Error generating neutral {email_type}"

def generate_simple_email(email_type, first_name, company_name=""):
    """Generate simple emails for Email 2 and Email 3"""
    template = prompt_templates.get(email_type, "")
    email = template.replace("[First Name]", first_name)
    email = email.replace("[COMPANY NAME]", company_name)
    return email.strip()

def process_row_single_email(row, row_index, email_type, client):
    """Process a single row for one specific email type"""
    try:
        # FIXED: Extract basic information with correct column indices
        first_name = str(row[0]).strip() if len(row) > 0 and row[0] else "Friend"  # Column A
        company_name = str(row[3]).strip() if len(row) > 3 and row[3] else "Your Company"  # Column D  
        website = str(row[6]).strip() if len(row) > 6 and row[6] else ""  # Column G
        industry = str(row[8]).strip() if len(row) > 8 and row[8] else ""  # Column I
        
        print(f"    👤 {first_name} from {company_name}")
        
        result = {}
        
        if email_type in SIMPLE_EMAIL_TYPES:
            # Handle Email 2 and Email 3 (simple templates)
            print(f"    📝 Generating simple email: {email_type}")
            email_content = generate_simple_email(email_type, first_name, company_name)
            result[email_type] = email_content
            print(f"    ✅ Generated {email_type}: {email_content[:50]}...")
            
        elif email_type in DATAPOINT_EMAIL_TYPES:
            # Handle emails that use datapoints
            print(f"    🔍 Collecting datapoints for {email_type}")
            datapoints = collect_datapoints(row)
            print(f"    📊 Found {len(datapoints)} valid datapoints")
            
            # Get the appropriate datapoint for this email type
            datapoint = get_datapoint_for_email_type(datapoints, email_type)
            
            if datapoint:
                print(f"    🎯 Using datapoint: {datapoint['category']} (rank {datapoint['rank']})")
                # Generate email with datapoint
                email_content = generate_email_with_datapoint(
                    email_type, 
                    datapoint['category'], 
                    datapoint['content'],
                    first_name,
                    company_name,
                    client
                )
                result[email_type] = email_content
                result[f"{email_type} Data Point"] = f"{datapoint['category']}: {datapoint['content'][:100]}..."
                print(f"    ✅ Generated {email_type} with datapoint: {email_content[:50]}...")
            else:
                print(f"    ⚠️  No suitable datapoint found, using neutral template")
                # Generate neutral email
                email_content = generate_neutral_email(
                    email_type, 
                    first_name, 
                    company_name,
                    website,
                    industry,
                    client
                )
                result[email_type] = email_content
                result[f"{email_type} Data Point"] = "No data point"
                print(f"    ✅ Generated neutral {email_type}: {email_content[:50]}...")
        
        print(f"    📋 Result keys: {list(result.keys())}")
        return result
    
    except Exception as e:
        print(f"    ❌ Error processing row {row_index + 1}: {e}")
        import traceback
        traceback.print_exc()
        return None

def update_sheet_single_email(service, batch_results, email_type):
    """Update the Google Sheet with results for a single email type"""
    try:
        data = []
        
        print(f"  🔍 Processing {len(batch_results)} results for {email_type}")
        
        for row_index, results in batch_results:
            if results is None:
                print(f"    ⚠️  Skipping row {row_index + 2} - no results")
                continue
                
            actual_row = row_index + 2  # Header is row 1, data starts from row 2
            print(f"    📝 Processing row {actual_row}, results keys: {list(results.keys()) if results else 'None'}")
            
            # Update the main email content
            if email_type in results:
                column_letter = output_columns[email_type]
                cell_range = f"{SHEET_NAME}!{column_letter}{actual_row}"
                email_content = results[email_type]
                print(f"      ✅ Adding {email_type} to {cell_range}: {email_content[:50]}...")
                data.append({
                    'range': cell_range,
                    'values': [[email_content]]
                })
            else:
                print(f"      ❌ {email_type} not found in results")
            
            # Update data point column if applicable
            datapoint_key = f"{email_type} Data Point"
            if datapoint_key in results and datapoint_key in output_columns:
                column_letter = output_columns[datapoint_key]
                cell_range = f"{SHEET_NAME}!{column_letter}{actual_row}"
                datapoint_content = results[datapoint_key]
                print(f"      ✅ Adding {datapoint_key} to {cell_range}: {datapoint_content[:50]}...")
                data.append({
                    'range': cell_range,
                    'values': [[datapoint_content]]
                })
        
        print(f"  📊 Total cells to update: {len(data)}")
        
        if data:
            body = {
                'valueInputOption': 'RAW',
                'data': data
            }
            
            print(f"  🚀 Executing batch update with {len(data)} cells...")
            result = service.spreadsheets().values().batchUpdate(
                spreadsheetId=SPREADSHEET_ID,
                body=body
            ).execute()
            
            updated_cells = result.get('totalUpdatedCells', 0)
            print(f"  ✅ Successfully updated {updated_cells} cells for {email_type}")
            return True
        else:
            print(f"  ⚠️  No data to update for {email_type}")
            return False
    
    except Exception as e:
        print(f"  ❌ Error updating {email_type}: {e}")
        import traceback
        traceback.print_exc()
        return False

def main_single_email(email_type, start_row=1):
    """Main function to run single email type generation
    
    Args:
        email_type (str): The email type to generate ('Email 1', 'Email 2', 'Subsequence 1', etc.)
        start_row (int): Row number to start processing from (1-based, excluding header)
    """
    print(f"🚀 Starting {email_type} generation from row {start_row}...")
    
    # Validate email type
    valid_email_types = ["Email 1", "Email 2", "Email 3", "Subsequence 1", "Subsequence 2", "Subsequence 3", "Subsequence 4"]
    
    if email_type not in valid_email_types:
        print(f"❌ Invalid email type: {email_type}")
        print(f"Valid options: {valid_email_types}")
        return
    
    # Initialize services
    sheets_service = authenticate_google_sheets()
    if not sheets_service:
        print("Failed to authenticate Google Sheets")
        return
    
    openai_client = None
    if email_type in DATAPOINT_EMAIL_TYPES:
        openai_client = initialize_openai()
        if not openai_client:
            print("Failed to initialize OpenAI client")
            return
    
    try:
        # Read data from Google Sheets
        range_name = f"{SHEET_NAME}!A:U"
        result = sheets_service.spreadsheets().values().get(
            spreadsheetId=SPREADSHEET_ID,
            range=range_name
        ).execute()
        
        values = result.get('values', [])
        if not values:
            print('No data found in the sheet.')
            return
        
        # Skip header row and validate start_row
        data_rows = values[1:]
        total_rows = len(data_rows)
        print(f"Found {total_rows} data rows in the sheet")
        
        if start_row < 1:
            print("Error: start_row must be >= 1")
            return
        elif start_row > total_rows:
            print(f"Error: start_row ({start_row}) is greater than total rows ({total_rows})")
            return
        
        # FIXED: Adjust for 0-based indexing
        start_index = start_row - 1
        remaining_rows = data_rows[start_index:]
        remaining_count = len(remaining_rows)
        print(f"Processing {remaining_count} rows for {email_type}")
        
        batch_size = 10
        successful_batches = 0
        
        # FIXED: Process rows in batches with correct indexing
        for batch_start in range(0, remaining_count, batch_size):
            batch_end = min(batch_start + batch_size, remaining_count)  # FIXED: Use remaining_count not total_rows
            batch_rows = remaining_rows[batch_start:batch_end]
            
            actual_start_row = start_row + batch_start
            actual_end_row = start_row + batch_end - 1
            print(f"\n📦 Processing batch {successful_batches + 1}: rows {actual_start_row} to {actual_end_row}")
            
            # Process each row in the current batch
            batch_results = []
            for i, row in enumerate(batch_rows):
                # FIXED: Correct row index calculation
                actual_row_index = start_index + batch_start + i
                print(f"  🔄 Processing row {actual_row_index + 2} (data row {actual_row_index + 1})")
                
                results = process_row_single_email(row, actual_row_index, email_type, openai_client)
                
                if results:
                    print(f"    ✅ Generated results: {list(results.keys())}")
                else:
                    print(f"    ❌ No results generated")
                
                batch_results.append((actual_row_index, results))
                
                # Add delay to avoid rate limiting
                if openai_client:  # Only delay if using OpenAI
                    import time
                    time.sleep(1)
            
            print(f"  📊 Batch has {len(batch_results)} results")
            
            # Update the sheet with this batch's results
            print(f"  🚀 Updating sheet for batch {successful_batches + 1}...")
            if update_sheet_single_email(sheets_service, batch_results, email_type):
                successful_batches += 1
                print(f"✅ Batch {successful_batches} completed successfully!")
            else:
                print(f"❌ Failed to update batch {successful_batches + 1}")
                print("Stopping process to prevent data loss...")
                break
        
        print(f"\n🎉 {email_type} generation completed! Successfully processed {successful_batches} batches.")
    
    except Exception as e:
        print(f"Error in main process: {e}")
        import traceback
        traceback.print_exc()

if __name__ == "__main__":
    # Example usage:
    # main_single_email("Email 2", start_row=1)
    # main_single_email("Subsequence 1", start_row=50)
    
    import sys
    if len(sys.argv) >= 2:
        email_type = 'Subsequence 4'
        start_row = 1
        main_single_email(email_type, start_row)
    else:
        print("Available email types: Email 1, Email 2, Email 3, Subsequence 1, Subsequence 2, Subsequence 3, Subsequence 4")
```
